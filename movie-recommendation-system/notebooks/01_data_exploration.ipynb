```python
{
 "cells": [
  {
   "cell_type": "markdown",
   "id": "8f1be32c",
   "metadata": {},
   "source": [
    "# 📊 Module 2: MovieLens Dataset Exploration & EDA\n",
    "**Movie Recommendation System Using Machine Learning**\n",
    "\n",
    "---\n",
    "\n",
    "### 🎯 Objectives\n",
    "1. Understand the structure and schema of the **MovieLens Latest Small** dataset.\n",
    "2. Load and inspect `movies.csv`, `ratings.csv`, `tags.csv`, and `links.csv` using Pandas.\n",
    "3. Compute statistical summaries and understand **Matrix Sparsity**.\n",
    "4. Visualize rating distributions, user activity patterns, and movie popularity curves.\n",
    "5. Practice beginner exercises to reinforce your learning.\n"
   ]
  },
  {
   "cell_type": "code",
   "execution_count": null,
   "id": "cc468d0e",
   "metadata": {},
   "outputs": [],
   "source": [
    "import os\n",
    "import pandas as pd\n",
    "import numpy as np\n",
    "import matplotlib.pyplot as plt\n",
    "import seaborn as sns\n",
    "\n",
    "# Set clean aesthetic styling\n",
    "sns.set_theme(style=\"whitegrid\")\n",
    "plt.rcParams[\"figure.figsize\"] = (9, 5)\n",
    "\n",
    "print(\"Environment ready!\")\n"
   ]
  },
  {
   "cell_type": "markdown",
   "id": "eb0df64e",
   "metadata": {},
   "source": [
    "## 1. Loading the Dataset\n",
    "We load the raw CSV files stored in the `dataset/` directory.\n",
    "- `movies.csv`: Maps each `movieId` to its `title` and pipe-separated `genres`.\n",
    "- `ratings.csv`: Contains historical user interactions `(userId, movieId, rating, timestamp)`.\n",
    "- `tags.csv`: Free-form user tags applied to movies.\n",
    "- `links.csv`: Identifiers linking to IMDb and TMDb.\n"
   ]
  },
  {
   "cell_type": "code",
   "execution_count": null,
   "id": "f9b32c50",
   "metadata": {},
   "outputs": [],
   "source": [
    "data_dir = \"../dataset\"\n",
    "\n",
    "movies_df = pd.read_csv(os.path.join(data_dir, \"movies.csv\"))\n",
    "ratings_df = pd.read_csv(os.path.join(data_dir, \"ratings.csv\"))\n",
    "tags_df = pd.read_csv(os.path.join(data_dir, \"tags.csv\"))\n",
    "links_df = pd.read_csv(os.path.join(data_dir, \"links.csv\"))\n",
    "\n",
    "print(f\"Movies Shape:  {movies_df.shape}\")\n",
    "print(f\"Ratings Shape: {ratings_df.shape}\")\n",
    "print(f\"Tags Shape:    {tags_df.shape}\")\n",
    "print(f\"Links Shape:   {links_df.shape}\")\n"
   ]
  },
  {
   "cell_type": "markdown",
   "id": "d5289cdf",
   "metadata": {},
   "source": [
    "## 2. Inspecting First Records and Data Types"
   ]
  },
  {
   "cell_type": "code",
   "execution_count": null,
   "id": "0fe401e1",
   "metadata": {},
   "outputs": [],
   "source": [
    "print(\"--- MOVIES HEAD ---\")\n",
    "display(movies_df.head(5))\n",
    "\n",
    "print(\"\\n--- RATINGS HEAD ---\")\n",
    "display(ratings_df.head(5))\n"
   ]
  },
  {
   "cell_type": "code",
   "execution_count": null,
   "id": "5340f436",
   "metadata": {},
   "outputs": [],
   "source": [
    "print(\"--- MOVIES INFO ---\")\n",
    "movies_df.info()\n",
    "\n",
    "print(\"\\n--- RATINGS INFO ---\")\n",
    "ratings_df.info()\n"
   ]
  },
  {
   "cell_type": "markdown",
   "id": "67623c02",
   "metadata": {},
   "source": [
    "## 3. Statistical Summaries & Catalog Sparsity\n",
    "\n",
    "### What is Matrix Sparsity?\n",
    "In collaborative filtering, users only rate a tiny fraction of available movies.\n",
    "$$\\text{Sparsity (\\%)} = \\left(1 - \\frac{\\text{Total Observed Ratings}}{\\text{Unique Users} \\times \\text{Unique Movies}}\\right) \\times 100$$\n"
   ]
  },
  {
   "cell_type": "code",
   "execution_count": null,
   "id": "85818cb1",
   "metadata": {},
   "outputs": [],
   "source": [
    "num_users = ratings_df[\"userId\"].nunique()\n",
    "num_movies = movies_df[\"movieId\"].nunique()\n",
    "num_ratings = len(ratings_df)\n",
    "\n",
    "total_possible = num_users * num_movies\n",
    "sparsity = (1.0 - (num_ratings / total_possible)) * 100.0\n",
    "\n",
    "print(f\"Unique Users:           {num_users:,}\")\n",
    "print(f\"Unique Movies:          {num_movies:,}\")\n",
    "print(f\"Total Observed Ratings: {num_ratings:,}\")\n",
    "print(f\"Total Possible Cells:   {total_possible:,}\")\n",
    "print(f\"Matrix Sparsity:        {sparsity:.2f}%\")\n",
    "print(f\"\\nRating Scale: min={ratings_df['rating'].min()}, max={ratings_df['rating'].max()}, mean={ratings_df['rating'].mean():.2f}\")\n"
   ]
  },
  {
   "cell_type": "markdown",
   "id": "e7953b87",
   "metadata": {},
   "source": [
    "## 4. Visualizing Rating Distributions\n",
    "Let's see how users distribute their scores (0.5 to 5.0).\n"
   ]
  },
  {
   "cell_type": "code",
   "execution_count": null,
   "id": "7278aa9f",
   "metadata": {},
   "outputs": [],
   "source": [
    "rating_counts = ratings_df[\"rating\"].value_counts().sort_index()\n",
    "\n",
    "plt.figure(figsize=(9, 5))\n",
    "ax = sns.barplot(x=rating_counts.index.astype(str), y=rating_counts.values, palette=\"Blues_d\")\n",
    "plt.title(\"Distribution of Ratings in MovieLens (100,836 ratings)\", fontsize=14, fontweight=\"bold\")\n",
    "plt.xlabel(\"Rating Score (Stars)\")\n",
    "plt.ylabel(\"Frequency\")\n",
    "\n",
    "for p in ax.patches:\n",
    "    h = p.get_height()\n",
    "    pct = (h / num_ratings) * 100\n",
    "    ax.annotate(f\"{pct:.1f}%\", (p.get_x() + p.get_width() / 2, h + 800), ha=\"center\", fontsize=9, fontweight=\"bold\")\n",
    "\n",
    "plt.show()\n"
   ]
  },
  {
   "cell_type": "markdown",
   "id": "f98dc38c",
   "metadata": {},
   "source": [
    "## 5. Identifying the Most-Rated Movies"
   ]
  },
  {
   "cell_type": "code",
   "execution_count": null,
   "id": "7c02b7e0",
   "metadata": {},
   "outputs": [],
   "source": [
    "top_movies = ratings_df[\"movieId\"].value_counts().head(10).reset_index()\n",
    "top_movies.columns = [\"movieId\", \"rating_count\"]\n",
    "top_movies = top_movies.merge(movies_df, on=\"movieId\")\n",
    "\n",
    "plt.figure(figsize=(10, 5))\n",
    "sns.barplot(data=top_movies, y=\"title\", x=\"rating_count\", palette=\"viridis\")\n",
    "plt.title(\"Top 10 Most-Rated Movies\", fontsize=14, fontweight=\"bold\")\n",
    "plt.xlabel(\"Number of Ratings\")\n",
    "plt.ylabel(\"Movie Title\")\n",
    "plt.show()\n"
   ]
  },
  {
   "cell_type": "markdown",
   "id": "ac42bbd9",
   "metadata": {},
   "source": [
    "## 6. Beginner Exercises & Common Pitfalls\n",
    "\n",
    "### 📝 Exercises\n",
    "1. Find the user who has rated the largest number of movies. How many ratings did they submit?\n",
    "2. Calculate the average rating for movies in the `Animation` genre versus `Action` genre.\n",
    "3. Why are half-star ratings (e.g. 0.5, 1.5, 2.5) less common than integer ratings (e.g. 3.0, 4.0)?\n",
    "\n",
    "### ⚠️ Common Errors & Fixes\n",
    "- **Error:** `KeyError: 'title'` when merging `ratings_df` and `movies_df`.\n",
    "  - **Fix:** Ensure merge is done using `on='movieId'` and that column names match casing.\n",
    "- **Error:** Memory crash when computing `ratings_df.pivot(...)` on massive datasets.\n",
    "  - **Fix:** Never pivot uncompressed millions of entries into dense matrices without sparse representation!\n"
   ]
  }
 ],
 "metadata": {},
 "nbformat": 4,
 "nbformat_minor": 5
}
```
